```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#used pandas for data analysis like seeing the first 5 lines of dataset using head() function etc.
#used matplot for charts and graphs
#used seaborn for visualization and understand the data patterns etc.
```

```python
iris = sns.load_dataset('iris')
iris.head()

#loaded the iris dataset and display it first 5 rows using head() function.
```

```python
print("Shape of dataset:", iris.shape)

#check the shape of dataset showing 150 rows and 5 columns
```

```python
print("Columns in dataset:")
print(iris.columns)

#displyed the column name in the dataset
```

```python
iris.describe()

#describe() function used to get the different things from the dataset like Standard deviation, min value, max value, mean etc.
```

```python
print(iris.isnull().sum())

#check the dataset if there is any missing value in any column using isnull() functn.
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title('Sepal Length vs Petal Length')
plt.show()

#each color is showing a different flower.
#we can see here a postive relation as when the petal length is increasing the sepal length is also increasing.
```

```python
plt.figure(figsize=(8,6))

sns.histplot(iris['sepal_length'], bins=20, kde=True)

plt.title('Distribution of Sepal Length')
plt.show()

#show the frequency of the sepal length and its distribution.
#most sepal length values are between 5 to 6.5
#curve is indicating a normal distribution
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(y=iris['petal_width'])

plt.title('Box Plot of Petal Width')
plt.show()

#used to detect if any outlier is there.
#it is showing an IQR the inter quartile range and there is no outlier present
```
